```python
import os #----> to add api key to environ variables
from langchain_groq import ChatGroq #----> to connect to groq llm models
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langchain_core.output_parsers import StrOutputParser #to access clean text from llm response object
```

```python
os.environ["GROQ_API_KEY"] = "your_groq_api_key_here"
# function to use any model of llm on groq
def use_model_from_groq(model_name:str,temperature:float):
    llm =  ChatGroq(
    model =model_name,
    temperature = temperature)
    return llm


```

## working of StrOutputParser

Prompt

  ↓
  
LLM

  ↓
  
AIMessage (object)

  ↓
  
StrOutputParser

  ↓
  
Plain string


```python
parser = StrOutputParser()
```

```python
## without ChatPromptTemplate
def chatbot():
    
    ##chat history
    chat_history = [
        SystemMessage(content = "you are an expert AI assistant and chatbot. give answers concisely and accuratelly")
        ]
    
    ##setting llm remote control
    llm =  use_model_from_groq("llama-3.3-70b-versatile",0.2)

        
            
    ##getting user input
    while True:
        given_question = input("enter what you want to ask:").strip()
        if given_question.lower()=="bye":
            break
        else:
            ##appending the message into chat_history
            chat_history.append(HumanMessage(content = given_question))
            chain = llm|parser
            response = chain.invoke(chat_history)
            chat_history.append(AIMessage(content = response))
            print(response)
            

chatbot()

    
        
    
    
    
```
